## Shabih Sazi Episode

```python
env = gym.make('CartPole-v1')
state = env.reset()

while True:
    action = env.action_space.sample()
    state, reward, terminated, truncated, info = env.step(action)
    print(action , reward)
    if terminated:
        break
env.close()
```

```python
env.render() #namayesh mohit va action haye baza
#dar mohit jupyter namayesh env azin tarigh emkan pazir nis
```

```python
from gym import wrappers # wrappers API k ghbeliat dastkari submodule haye gym ra b ma midahad
```

```python
env = gym.make('CartPole-v1' , render_mode='rgb_array')
env = wrappers.RecordVideo(env, 'videos')
state = env.reset()

while True:
    action = env.action_space.sample()
    state, reward, terminated, truncated, info = env.step(action)
    print(action , reward)
    if terminated:
        break
env.close()
```
